## El problema: qué hacer con el etanol de la precipitación

En la planta de pectina, la pectina se recupera del extracto clarificado
precipitándola con etanol. El licor madre que sale del filtro es una mezcla de
agua y etanol con algo de azúcares y ácidos disueltos, y su volumen no es
pequeño: por cada tonelada de cáscara procesada se manejan varios metros cúbicos
de mezcla hidroalcohólica.

Ese licor no se puede botar y no se puede reutilizar tal como sale. No se puede
botar porque el etanol es el insumo más caro del proceso y porque una corriente
con 45 % de alcohol en volumen no entra a un tratamiento biológico. Y no se puede
reutilizar porque la precipitación necesita una concentración alta de etanol para
funcionar: si el solvente vuelve diluido, la pectina no precipita.

La respuesta industrial es una **columna de destilación continua** que parte el
licor madre en dos: por la cabeza sale solvente concentrado que vuelve al tanque
de precipitación, y por el fondo sale una corriente acuosa con poco etanol que ya
puede ir a tratamiento. Este taller dimensiona esa columna.

### Las especificaciones de la columna

| Corriente | Fracción molar de etanol | Equivalente en volumen | De dónde sale o a dónde va |
|---|---|---|---|
| Alimentación, licor madre | $x_F = 0{,}20$ | 44,7 % v/v | Filtro de la precipitación |
| Destilado, solvente recuperado | $x_D = 0{,}80$ | 92,8 % v/v | Tanque de precipitación |
| Fondos, efluente acuoso | $x_B = 0{,}02$ | 6,2 % v/v | Tratamiento de aguas |

La alimentación llega **subenfriada**, porque viene del filtro y no de un
precalentador: $q = 1{,}3$. La relación de reflujo de diseño es $R = 2{,}5$.

## Marco teórico

### Qué representa una columna de destilación

Una columna de destilación se modela como una serie de etapas de contacto entre
líquido y vapor. En cada etapa el vapor que asciende entra en contacto con el
líquido que desciende. Si la etapa es **ideal**, las dos corrientes que salen
están en equilibrio termodinámico.

En la cabeza está el **condensador**. Parte del líquido condensado se retira como
destilado y el resto vuelve a la columna como **reflujo**. En el fondo está el
**rehervidor** o calderín, que suministra el calor que vaporiza parte del líquido
de fondos y sostiene el flujo de vapor ascendente.

La columna queda partida en dos zonas por el plato de alimentación:

- **Zona de rectificación**, por encima de la alimentación. El vapor se enriquece
  en el componente más volátil.
- **Zona de agotamiento**, por debajo. El líquido se empobrece en el componente
  más volátil.

El método de McCabe-Thiele representa cada zona con su propia línea de operación.

### El diagrama de equilibrio

El método se construye sobre el diagrama $x$ contra $y$, donde $x$ es la fracción
molar del componente más volátil en el líquido y $y$ la del mismo componente en
el vapor. En esta columna el componente más volátil es el etanol.

La **curva de equilibrio** da la composición del vapor que está en equilibrio con
cada composición líquida. Cuando el componente elegido es el más volátil se
cumple $y > x$, así que la curva queda por encima de la diagonal $y = x$.

La diagonal no es una línea de operación. Es la referencia que hace funcionar la
construcción gráfica, que alterna tramos horizontales hacia la curva de
equilibrio con tramos verticales hacia las líneas de operación.

Cuando la curva de equilibrio se acerca a la diagonal, la separación se vuelve
difícil: el vapor apenas se enriquece respecto al líquido y hacen falta muchas
etapas para avanzar poco.

### El azeótropo

Algunas mezclas no se comportan de forma ideal y su curva de equilibrio llega a
**cruzar** la diagonal. En el punto de cruce,

$$x = y$$

y la mezcla hierve como si fuera una sustancia pura. Ese punto es un
**azeótropo**, y la destilación convencional no lo puede atravesar.

El sistema etanol-agua a presión atmosférica tiene un azeótropo en fracción molar
de etanol igual a 0,894, que corresponde a 95,6 % en masa. Por eso el alcohol
antiséptico se vende a 96 % y no a 100 %. Para pasar de ahí hacen falta otras
rutas: destilación azeotrópica, destilación extractiva, tamices moleculares,
cambio de presión o pervaporación.

Este taller trabaja con el azeótropo **dentro** de la tabla de equilibrio, no
fuera de ella. Va a aparecer en los cálculos, no solo en el texto.

### Línea de operación de rectificación

Del balance de materia en la zona por encima de la alimentación sale

$$y = \frac{R}{R+1}\,x + \frac{x_D}{R+1}$$

La recta pasa por el punto $(x_D, x_D)$ sobre la diagonal y tiene pendiente
$R/(R+1)$. La **relación de reflujo** es

$$R = \frac{L}{D}$$

con $L$ el líquido que retorna a la columna y $D$ el destilado retirado.

Subir el reflujo mejora la separación porque aumenta el contacto interno, pero
cuesta energía en el condensador y en el rehervidor. Bajarlo ahorra energía y
exige más etapas. En el límite $R \to \infty$ se llega al **reflujo total**, donde
las dos líneas de operación coinciden con la diagonal y el número de etapas es el
mínimo posible, pero no se produce destilado.

### Línea de alimentación

La condición térmica de la alimentación se resume en el parámetro $q$, la
fracción de la alimentación que se comporta como líquido al entrar. La línea $q$
es

$$y = \frac{q}{q-1}\,x - \frac{x_F}{q-1}$$

y su interpretación física es directa:

| $q$ | Estado de la alimentación | Línea $q$ |
|---|---|---|
| $q > 1$ | Líquido subenfriado | Pendiente positiva mayor que 1 |
| $q = 1$ | Líquido saturado | Vertical en $x = x_F$ |
| $0 < q < 1$ | Mezcla líquido-vapor | Pendiente negativa |
| $q = 0$ | Vapor saturado | Horizontal |
| $q < 0$ | Vapor sobrecalentado | Pendiente entre 0 y 1 |

Note que en $q = 1$ la ecuación se indetermina y hay que tratar el caso aparte.
Esa bifurcación es la primera trampa de la implementación.

### Línea de operación de agotamiento

La zona por debajo de la alimentación se representa con la recta que pasa por dos
puntos conocidos: el de fondos, $(x_B, x_B)$, y la intersección de la LOR con la
línea $q$, $(x_{int}, y_{int})$. Con esos dos puntos,

$$m = \frac{y_{int} - x_B}{x_{int} - x_B}, \qquad b = x_B\,(1 - m), \qquad y = m x + b$$

### Etapas teóricas

Una **etapa teórica** es una etapa ideal de contacto en la que las corrientes que
salen están en equilibrio. En una columna real una etapa puede ser un plato, un
tramo de empaque o una zona de contacto, y ninguna es ideal, así que después hay
que convertir etapas teóricas en platos reales con una eficiencia.

La construcción es la siguiente:

1. Partir de $(x_D, x_D)$.
2. Avanzar horizontalmente hasta la curva de equilibrio. Ese tramo es una etapa.
3. Bajar verticalmente hasta la línea de operación que corresponda.
4. Cambiar de la LOR a la LOA al pasar por $x_{int}$.
5. Repetir hasta alcanzar $x_B$.

El número de tramos es el número de etapas teóricas. El **plato de alimentación**
es la etapa en la que se produce el cambio de línea.

Dos detalles que la mayoría de los textos pasa por alto y que aquí importan:

- **El rehervidor cuenta como una etapa.** Si el escalonado da $N$ etapas, la
  columna lleva $N-1$ platos y el calderín aporta la que falta.
- **La última etapa casi nunca cae exacta sobre $x_B$.** Se pasa. Reportar el
  entero de arriba infla el resultado, así que se informa la **fracción** de la
  última etapa, interpolando linealmente entre el punto anterior y el que se
  pasó.

```python
import re

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.optimize import brentq, fsolve
from scipy.interpolate import PchipInterpolator

# ----------------------------------------------------------------------
# Coma decimal: la convención del curso es coma para el decimal y punto
# para el millar. Se aplica sobre el texto ya formateado, así que sirve
# igual para una f-string y para una tabla de pandas.
#
# Lo que NO se toca: la numeración de secciones. En «Tarea 2.4» ese punto
# no es un decimal, y cambiarlo dejaría títulos como «Tarea 2,4».
# ----------------------------------------------------------------------
_ROTULO = re.compile(
    r'(?i)(tarea|parte|pregunta|caso|ejercicio|figura|tabla|secci[oó]n|paso|'
    r'inciso|cap[ií]tulo|nota)\s+$'
)


def es(texto):
    """Devuelve el texto con la coma decimal del español."""
    t = str(texto)

    def cambia(m):
        if _ROTULO.search(t[max(0, m.start() - 16):m.start()]):
            return m.group(0)
        return m.group(0).replace('.', ',')

    return re.sub(r'\d+\.\d+', cambia, t)


# Las tablas que se muestran con display() no pasan por print, así que la coma
# se les pone por el formateador de pandas.
pd.set_option('display.float_format', lambda v: f'{v:.4f}'.replace('.', ','))

plt.rcParams.update({'figure.dpi': 110, 'font.size': 11,
                     'axes.grid': True, 'grid.linestyle': ':',
                     'grid.alpha': 0.55, 'axes.axisbelow': True})

print(es('Librerías cargadas. Coma decimal activa: pi = %.4f' % np.pi))
```

```python
# Equilibrio etanol-agua a 1 atm. Datos experimentales, con el azeótropo dentro.
#
# La última columna es la que manda en este taller: y - x mide cuánto enriquece
# el vapor respecto al líquido. Donde vale cero no hay separación posible.

x_EtOH = np.array([0.000, 0.010, 0.020, 0.050, 0.100, 0.200, 0.300, 0.400,
                   0.500, 0.550, 0.650, 0.750, 0.850, 0.894, 0.950, 1.000])
y_EtOH = np.array([0.000, 0.110, 0.175, 0.325, 0.442, 0.531, 0.576, 0.614,
                   0.654, 0.676, 0.722, 0.778, 0.856, 0.894, 0.942, 1.000])

X_AZ = 0.894   # azeótropo etanol-agua a 1 atm, 95,6 % en masa

tabla = pd.DataFrame({'x': x_EtOH, 'y': y_EtOH, 'y - x': y_EtOH - x_EtOH})
print(es(tabla.to_string(index=False)))
print()
i_az = int(np.argmin(np.abs(x_EtOH - X_AZ)))
print(es(f'El azeótropo está en x = {X_AZ:.3f}, donde y - x vale '
         f'{y_EtOH[i_az] - x_EtOH[i_az]:+.3f}'))
print(es('Por encima de ese punto y < x: los papeles se invierten y el agua '
         'pasa a ser el componente que se enriquece en el vapor.'))
```

```python
# Interpolación monótona de la curva de equilibrio.
#
# Se usa PCHIP y no un spline cúbico corriente porque PCHIP no inventa
# oscilaciones entre los puntos. Un spline puede devolver y > 1 o cruzar la
# diagonal donde los datos no lo hacen, y eso rompe el escalonado.
#
# La inversa se puede construir directamente porque y crece de forma monótona
# con x en todo el intervalo. Ojo: un azeótropo es donde y = x, NO donde y deja
# de crecer. La curva sigue siendo monótona al pasar por el azeótropo.

interp_xy = PchipInterpolator(x_EtOH, y_EtOH)
interp_yx = PchipInterpolator(y_EtOH, x_EtOH)

vapor_eq = lambda x: float(interp_xy(x))
liquido_eq = lambda y: float(interp_yx(y))

print(es(f'monótona creciente: {bool(np.all(np.diff(y_EtOH) > 0))}'))
for v in (0.10, 0.50, 0.80, X_AZ):
    print(es(f'  y({v:.3f}) = {vapor_eq(v):.4f}   y - x = {vapor_eq(v) - v:+.4f}'))

cruce = brentq(lambda x: vapor_eq(x) - x, 0.85, 0.99)
print(es(f'\nCruce de la diagonal por búsqueda de raíz: x = {cruce:.4f}'))
```

```python
def linea_rectificacion(R, x_D):
    """Devuelve la LOR como función de x."""
    return lambda x: (R / (R + 1)) * x + x_D / (R + 1)


def interseccion_q_lor(R, x_D, q, x_F):
    """Punto donde la línea q corta la LOR.

    El caso q = 1 se trata aparte: la línea q es vertical y la ecuación
    general se indetermina.
    """
    lor = linea_rectificacion(R, x_D)
    if abs(q - 1.0) < 1e-9:
        x_i = x_F
    else:
        lq = lambda x: (q / (q - 1)) * x - x_F / (q - 1)
        x_i = brentq(lambda x: lq(x) - lor(x), -5.0, 5.0)
    return x_i, lor(x_i)


def linea_agotamiento(x_B, x_int, y_int):
    """Devuelve la LOA, la recta por (x_B, x_B) y (x_int, y_int)."""
    m = (y_int - x_B) / (x_int - x_B)
    b = x_B * (1 - m)
    return lambda x: m * x + b


# Caso de demostración
xF_d, xD_d, xB_d, q_d, R_d = 0.24, 0.70, 0.05, 1.0, 2.0

lor_d = linea_rectificacion(R_d, xD_d)
xi_d, yi_d = interseccion_q_lor(R_d, xD_d, q_d, xF_d)
loa_d = linea_agotamiento(xB_d, xi_d, yi_d)

print(es(f'Pendiente de la LOR: R/(R+1) = {R_d / (R_d + 1):.4f}'))
print(es(f'Intersección con la línea q: ({xi_d:.4f} ; {yi_d:.4f})'))
print(es(f'Pendiente de la LOA: {(yi_d - xB_d) / (xi_d - xB_d):.4f}'))
print(es('La LOA es más empinada que la diagonal, como debe ser en la zona '
         'de agotamiento.'))
```

```python
def escalonar(x_D, x_B, R, q, x_F, liquido_eq, tope=200):
    """Construye los escalones de McCabe-Thiele desde x_D hasta x_B.

    Devuelve un diccionario con la traza para graficar, el número de etapas
    con su fracción final, el plato de alimentación y un aviso de no
    convergencia.

    Tres cosas que el algoritmo ingenuo hace mal y aquí no:

    1. **Tope de iteraciones.** Si el reflujo queda por debajo del mínimo, las
       líneas de operación tocan la curva de equilibrio y los escalones se
       hacen infinitamente pequeños sin llegar nunca a x_B. Sin tope, el bucle
       no termina.
    2. **Fracción de la última etapa.** El último escalón se pasa de x_B. Se
       interpola para no inflar el resultado.
    3. **Cambio de línea por x_int, no por x_F.** El cruce de zona ocurre en la
       intersección de las líneas de operación, que en general no coincide con
       la composición de la alimentación.
    """
    lor = linea_rectificacion(R, x_D)
    x_int, y_int = interseccion_q_lor(R, x_D, q, x_F)
    loa = linea_agotamiento(x_B, x_int, y_int)

    xs, ys = [x_D], [x_D]
    x_ant, y_ant = x_D, x_D
    n_enteras, plato_alim = 0, None

    while x_ant > x_B and n_enteras < tope:
        # tramo horizontal: del punto de operación a la curva de equilibrio
        x_nue = liquido_eq(y_ant)
        xs.append(x_nue)
        ys.append(y_ant)
        n_enteras += 1

        if x_nue <= x_B:            # esta etapa ya se pasó del fondo
            break

        # tramo vertical: de la curva de equilibrio a la línea de operación
        if x_nue > x_int:
            y_nue = lor(x_nue)
        else:
            if plato_alim is None:
                plato_alim = n_enteras
            y_nue = loa(x_nue)

        xs.append(x_nue)
        ys.append(y_nue)
        x_ant, y_ant = x_nue, y_nue

    # fracción de la última etapa
    if x_nue < x_B < x_ant:
        fraccion = (x_ant - x_B) / (x_ant - x_nue)
        N = (n_enteras - 1) + fraccion
    else:
        fraccion, N = 1.0, float(n_enteras)

    return {'xs': xs, 'ys': ys, 'N': N, 'N_enteras': n_enteras,
            'fraccion': fraccion, 'plato_alim': plato_alim,
            'x_int': x_int, 'y_int': y_int, 'x_final': x_nue,
            'no_convergio': n_enteras >= tope}


r_d = escalonar(xD_d, xB_d, R_d, q_d, xF_d, liquido_eq)
print(es(f'Etapas teóricas: {r_d["N"]:.2f}  '
         f'({r_d["N_enteras"] - 1} completas más {r_d["fraccion"]:.2f} de la última)'))
print(es(f'Plato de alimentación: etapa {r_d["plato_alim"]}'))
print(es(f'El escalonado terminó en x = {r_d["x_final"]:.4f}, '
         f'con x_B = {xB_d:.3f}'))
print(es(f'Redondeando hacia arriba: {int(np.ceil(r_d["N"]))} etapas, '
         f'{int(np.ceil(r_d["N"])) - 1} platos más el calderín'))
```

```python
def diagrama(res, x_D, x_B, R, q, x_F, vapor_eq, titulo, componente='etanol',
             marcar_azeotropo=None):
    """Diagrama de McCabe-Thiele completo, con escalones y plato de alimentación."""
    fig, ax = plt.subplots(figsize=(7.4, 7.4))
    xg = np.linspace(0, 1, 400)

    ax.plot(xg, [vapor_eq(v) for v in xg], color='#1f4e79', lw=2.4,
            label='Curva de equilibrio', zorder=3)
    ax.plot([0, 1], [0, 1], color='#666', lw=1.2, ls='--', label='Diagonal $y=x$')

    # línea q
    if abs(q - 1.0) < 1e-9:
        ax.axvline(x_F, color='#7b3f9d', lw=1.6, ls=':', label=f'Línea q, q = {q:g}')
    else:
        yq = (q / (q - 1)) * xg - x_F / (q - 1)
        dentro = (yq >= -0.02) & (yq <= 1.02)
        ax.plot(xg[dentro], yq[dentro], color='#7b3f9d', lw=1.6, ls=':',
                label=f'Línea q, q = {q:g}')

    lor = linea_rectificacion(R, x_D)
    loa = linea_agotamiento(x_B, res['x_int'], res['y_int'])
    xr = np.linspace(res['x_int'], x_D, 60)
    xs_ = np.linspace(x_B, res['x_int'], 60)
    ax.plot(xr, [lor(v) for v in xr], color='#c0392b', lw=1.8, label=f'LOR, R = {R:g}')
    ax.plot(xs_, [loa(v) for v in xs_], color='#1e8449', lw=1.8, label='LOA')

    # Las etiquetas se arman antes de la llamada: anidar f-strings con el mismo
    # tipo de comilla solo es válido desde Python 3.12, y Colab no siempre lo es.
    rot_N = es('%.2f' % res['N'])
    ax.plot(res['xs'], res['ys'], color='#e67e22', lw=1.5,
            label=f'Escalones, N = {rot_N}', zorder=4)

    ax.plot([x_D], [x_D], 'o', color='#c0392b', ms=7, zorder=5)
    ax.plot([x_B], [x_B], 'o', color='#1e8449', ms=7, zorder=5)
    ax.plot([x_F], [x_F], 's', color='#7b3f9d', ms=7, zorder=5)
    ax.plot([res['x_int']], [res['y_int']], 'ko', ms=5, zorder=5)

    # El punto de la etapa k está en el índice 2k-1 de la traza, porque la
    # lista arranca con (x_D, x_D) y cada etapa añade dos puntos.
    k = res['plato_alim']
    if k is not None and 2 * k - 1 < len(res['xs']):
        ax.plot([res['xs'][2 * k - 1]], [res['ys'][2 * k - 1]], '*',
                color='#c0392b', ms=18, zorder=6,
                label=f'Plato de alimentación, etapa {k}')

    if marcar_azeotropo is not None:
        rot_az = es('%.3f' % marcar_azeotropo)
        ax.plot([marcar_azeotropo], [marcar_azeotropo], 'X', color='#000',
                ms=10, zorder=6, label=f'Azeótropo, x = {rot_az}')

    ax.set_xlabel(f'x, fracción molar de {componente} en el líquido')
    ax.set_ylabel(f'y, fracción molar de {componente} en el vapor')
    ax.set_title(titulo)
    ax.set_xlim(0, 1)
    ax.set_ylim(0, 1)
    ax.set_aspect('equal')
    ax.legend(loc='lower right', fontsize=9, framealpha=0.94)
    plt.tight_layout()
    plt.show()


diagrama(r_d, xD_d, xB_d, R_d, q_d, xF_d, vapor_eq,
         'Demostración: recuperación de etanol a 0,70 de fracción molar',
         marcar_azeotropo=X_AZ)
```

## Lo que la demostración deja ver

Tres cosas para mirar en la figura antes de seguir.

**La curva de equilibrio del etanol-agua no se parece a la de un sistema ideal.**
Sube muy rápido en la zona diluida, de 0 a 0,44 mientras $x$ pasa de 0 a 0,10, y
después casi se aplana. Ese codo es lo que hace que la separación sea fácil al
principio y muy difícil al final.

**Los escalones no tienen el mismo tamaño.** Los primeros, cerca del destilado,
son estrechos. Los últimos, en la zona diluida, son enormes: uno solo baja de
0,054 a 0,004 de fracción molar. La eficiencia por etapa depende de dónde esté la
etapa.

**El azeótropo está marcado y está cerca.** El destilado de la demostración,
0,70, deja margen. El de diseño, 0,80, deja mucho menos, y el caso de estudio 2
va a mostrar que ese margen es lo que gobierna el costo de la columna.

Una advertencia sobre modelos: puede tentarle representar el equilibrio con
volatilidad relativa constante, $y = \alpha x / (1 + (\alpha - 1)x)$, que es
cómodo porque tiene inversa analítica. Para el etanol-agua eso no funciona. Con
la $\alpha$ que ajusta el tramo diluido, el modelo predice $y = 0{,}90$ donde el
dato real es $0{,}65$, y la columna sale con la mitad de las etapas que necesita.
La volatilidad relativa de este sistema pasa de cerca de 12 en dilución a
exactamente 1 en el azeótropo. No hay un valor constante que sirva.

### TODO 2 — Reflujo mínimo (13 puntos)

El reflujo mínimo es la frontera de lo posible: por debajo de él, ninguna cantidad

```python
# Tarea 3.3 (4 puntos)
#
# Modelo de Van Laar del isopropanol-agua, ajustado al azeótropo real del
# sistema. Es el mismo modelo del taller 8. Ejecute esta celda tal como está.

ANT_VL = {'ipa': (8.87829, 2010.33, 252.636), 'agua': (8.07131, 1730.63, 233.426)}
XAZ_IPA, TAZ_IPA = 0.685, 80.4


def _psat_vl(comp, T):
    A, B, C = ANT_VL[comp]
    return 10 ** (A - B / (C + T))


def _gammas(x1, A, B):
    x2 = 1 - x1
    if x1 <= 0:
        return float(np.exp(A)), 1.0
    if x2 <= 0:
        return 1.0, float(np.exp(B))
    return (float(np.exp(A / (1 + (A * x1) / (B * x2)) ** 2)),
            float(np.exp(B / (1 + (B * x2) / (A * x1)) ** 2)))


def _ecuaciones(p):
    A, B = p
    g1, g2 = _gammas(XAZ_IPA, A, B)
    return [g1 * _psat_vl('ipa', TAZ_IPA) - g2 * _psat_vl('agua', TAZ_IPA),
            g1 * XAZ_IPA * _psat_vl('ipa', TAZ_IPA)
            + g2 * (1 - XAZ_IPA) * _psat_vl('agua', TAZ_IPA) - 760.0]


A_VL, B_VL = fsolve(_ecuaciones, [2.4, 1.1])


def y_real_ipa(x1):
    """Equilibrio real del isopropanol-agua a 760 mmHg, Van Laar."""
    if x1 <= 0:
        return 0.0
    if x1 >= 1:
        return 1.0

    def f(T):
        g1, g2 = _gammas(x1, A_VL, B_VL)
        return (g1 * x1 * _psat_vl('ipa', T)
                + g2 * (1 - x1) * _psat_vl('agua', T) - 760.0)

    T = brentq(f, 40.0, 140.0)
    g1, _ = _gammas(x1, A_VL, B_VL)
    return g1 * x1 * _psat_vl('ipa', T) / 760.0


print(es(f'Van Laar ajustado: A = {A_VL:.4f}   B = {B_VL:.4f}'))
print(es(f'Comprobación del azeótropo: y({XAZ_IPA:.3f}) = '
         f'{y_real_ipa(XAZ_IPA):.4f}'))

# ESCRIBA SU CÓDIGO AQUÍ: las dos curvas, la tabla de diferencias y el
# diagnóstico de la columna de la tarea 3.2.
```
