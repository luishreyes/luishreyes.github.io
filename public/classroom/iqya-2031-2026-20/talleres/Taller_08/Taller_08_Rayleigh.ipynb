## Introducción y Marco Teórico

### ¿Qué es la Destilación en Discontinuo?

La **destilación batch** o **destilación en discontinuo** es una de las operaciones unitarias de separación más antiguas y, al mismo tiempo, una de las más útiles para comprender el comportamiento real de una mezcla líquida durante una separación por vaporización. A diferencia de una columna de destilación continua, en la que la alimentación entra de manera permanente y las corrientes de producto se retiran en régimen estacionario, en una destilación batch se carga inicialmente una cantidad finita de mezcla en un calderín y luego se calienta hasta generar vapor. Ese vapor se retira, se condensa y se recolecta como destilado.

Durante el proceso, la composición del líquido que permanece en el calderín cambia continuamente. Si se desea recuperar el **componente más volátil** (CMV), el vapor instantáneo suele estar enriquecido en ese componente respecto al líquido. Por lo tanto, a medida que avanza la operación, el líquido remanente se va **empobreciendo** en el CMV y el destilado acumulado representa una mezcla de todos los vapores generados desde el inicio hasta el final del proceso.

En términos generales, una destilación batch se caracteriza porque:

- Se carga una mezcla líquida binaria o multicomponente en un calderín.
- El sistema se calienta hasta ebullición y el vapor generado se retira continuamente.
- El vapor se asume en equilibrio con el líquido del calderín en cada instante.
- La composición del líquido cambia con el tiempo, por lo que el proceso no es estacionario.
- El proceso se detiene cuando se alcanza una composición final deseada, una cantidad destilada determinada o una especificación de pureza.

Esta operación aparece con frecuencia en la **industria farmacéutica**, en la recuperación de solventes, en plantas piloto, en la producción de fragancias, en laboratorios de investigación, en procesos de purificación de productos de alto valor agregado y en la producción tradicional de bebidas espirituosas. Su flexibilidad la hace especialmente útil cuando los lotes son pequeños, cuando se procesan varias mezclas en el mismo equipo o cuando no se justifica económicamente una columna continua.

### Importancia histórica del método de Rayleigh

El análisis matemático de la destilación diferencial se asocia comúnmente con la **ecuación de Rayleigh**, desarrollada a partir de balances diferenciales de materia aplicados a sistemas en los que el vapor se retira de manera continua desde un líquido en ebullición. Lord Rayleigh realizó contribuciones fundamentales en física, termodinámica y fenómenos de transporte, y su nombre aparece en diferentes áreas de la ingeniería y la ciencia. En el contexto de la destilación, la ecuación de Rayleigh permite relacionar el cambio de composición del líquido remanente con la cantidad de material que permanece en el calderín.

La utilidad de esta ecuación radica en que convierte un proceso no estacionario en un problema integrable, siempre que se conozca la relación de equilibrio líquido-vapor $y^* = f(z_L)$. ### Suposiciones del Modelo

Para derivar y aplicar la ecuación de Rayleigh se asumen las siguientes condiciones:

1. **Equilibrio termodinámico instantáneo**: el vapor que abandona el calderín en cada instante está en equilibrio con el líquido, por lo que $y^* = f(z_L)$.
2. **Vaporización diferencial**: en cada paso se vaporiza una cantidad infinitesimal de líquido.
3. **Sin retención significativa de vapor**: la fase vapor sale del sistema tan pronto se forma.
4. **Mezcla perfecta en el calderín**: la composición del líquido es uniforme en todo momento.
5. **Presión aproximadamente constante**: la curva de equilibrio utilizada corresponde a la presión de operación del sistema.
6. **Sin pérdidas externas de materia**: todo el material que sale del calderín se recolecta como destilado.

Estas suposiciones simplifican el análisis y permiten concentrarse en la relación entre equilibrio, composición y cantidad de líquido remanente. En equipos reales pueden existir pérdidas de calor, holdup de vapor, eficiencia limitada, arrastre de gotas y desviaciones respecto al equilibrio, pero el modelo de Rayleigh sigue siendo una herramienta fundamental para el análisis preliminar.

## Derivación de la Integral de Separación

### Balance diferencial de componente

Consideramos el calderín como volumen de control. En un instante $dt$ se vaporiza $dL$ moles con composición $y^*$:

$$d(L \cdot z_L) = -y^* \cdot dL$$

Expandiendo el diferencial del lado izquierdo:

$$L \cdot dz_L + z_L \cdot dL = -y^* \cdot dL$$

Reorganizando y separando variables:

$$\frac{dL}{L} = \frac{dz_L}{z_L - y^*}$$

### Integración entre estado inicial y final

$$\int_{L_0}^{L_f} \frac{dL}{L} = \int_{z_0}^{z_f} \frac{dz_L}{z_L - y^*}$$

$$\ln\left(\frac{L_f}{L_0}\right) = \int_{z_0}^{z_f} \frac{dz_L}{z_L - y^*}$$

Como $z_f < z_0$ (el líquido se empobrece) y $y^* > z_L$ (el vapor es más rico en CMV que el líquido), la integral es negativa cuando se evalúa en ese sentido. Invirtiendo los límites:

### Integral de Separación (forma de Rayleigh)

$$\boxed{\ln\left(\frac{L_0}{L_f}\right) = \int_{z_f}^{z_0} \frac{dz_L}{y^*(z_L) - z_L}}$$

Esta integral es el corazón del modelo. Su valor determina la razón $L_0/L_f$ y, por ende, la fracción de líquido vaporizada.

### Modelos de Equilibrio Vapor-Líquido

#### Modelo de Volatilidad Relativa Constante

Para sistemas que se comportan de forma aproximadamente ideal (o cuando se dispone de $\alpha$ promedio):

$$y^* = \frac{\alpha \cdot z}{1 + (\alpha - 1) \cdot z}$$

Este modelo es exacto para mezclas que siguen la Ley de Raoult y $P^{sat}$ sigue la ecuación de Antoine.

```python
def equilibrio_alpha_cte(z, alpha):
    """
    Calcula la composición del vapor en equilibrio (y*) usando volatilidad relativa constante.

    Parámetros
    ----------
    z     : float o array  →  fracción molar del CMV en el líquido
    alpha : float          →  volatilidad relativa (>1 para que haya separación)

    Retorna
    -------
    y_star : float o array  →  fracción molar del CMV en el vapor
    """
    z = np.clip(z, 0.0, 1.0)                # Asegurar rango físico
    y_star = (alpha * z) / (1.0 + (alpha - 1.0) * z)
    return y_star

# ── Comprobación rápida ──
alpha_prueba = 2.8
print(es("Verificación del modelo de equilibrio (α = 2.8):"))
print(es(f"  z=0.0  → y* = {equilibrio_alpha_cte(0.0, alpha_prueba):.4f}  (esperado: 0.0000)"))
print(es(f"  z=0.5  → y* = {equilibrio_alpha_cte(0.5, alpha_prueba):.4f}  (esperado: 0.7368)"))
print(es(f"  z=1.0  → y* = {equilibrio_alpha_cte(1.0, alpha_prueba):.4f}  (esperado: 1.0000)"))
```

#### Modelo con Datos Experimentales (Interpolación)

Cuando se dispone de pares $(x_i, y_i)$ medidos experimentalmente, se construye un interpolador cúbico o lineal.

```python
def construir_equilibrio_experimental(x_exp, y_exp, tipo='cubica'):
    """
    Construye una función de equilibrio a partir de datos experimentales discretos.

    Parámetros
    ----------
    x_exp : array  →  composiciones líquido medidas
    y_exp : array  →  composiciones vapor medidas (en equilibrio con x_exp)
    tipo  : str    →  'cubica', 'lineal' o 'cuadratica'

    Retorna
    -------
    f_eq  : función  →  f_eq(z) devuelve y*(z) interpolado
    """
    idx = np.argsort(x_exp)
    x_s = np.array(x_exp)[idx]
    y_s = np.array(y_exp)[idx]

    kind_map = {'cubica': 'cubic', 'lineal': 'linear', 'cuadratica': 'quadratic'}
    kind = kind_map.get(tipo, 'cubic')

    f_eq = interp1d(x_s, y_s, kind=kind, bounds_error=False, fill_value='extrapolate')
    return f_eq

print(es("Funciones de equilibrio definidas"))
```

### Evaluación Numérica de la Integral de Separación

La integral

$$I = \int_{z_f}^{z_0} \frac{dz}{y^*(z) - z}$$

se evalúa numéricamente. Se ofrecen tres esquemas: cuadratura adaptativa de Gauss-Kronrod (`quad`), regla del trapecio y regla de Simpson.

```python
def calcular_integral_separacion(z0, zf, func_equilibrio, esquema='quad'):
    """
    Evalúa la integral de separación de Rayleigh numéricamente.

    Parámetros
    ----------
    z0              : float   →  composición inicial del líquido
    zf              : float   →  composición final del líquido  (zf < z0)
    func_equilibrio : función →  devuelve y*(z)
    esquema         : str     →  'quad' | 'trapz' | 'simpson'

    Retorna
    -------
    dict con claves:
        'integral'           : valor de I
        'razon_L0_Lf'        : L0/Lf = exp(I)
        'fraccion_vaporizada': (L0 - Lf)/L0
        'fraccion_liquida'   : Lf/L0
        'error_estimado'     : error de la cuadratura (solo para 'quad')
    """
    if zf >= z0:
        raise ValueError(f"Se requiere zf ({zf:.4f}) < z0 ({z0:.4f})")

    def integrando(z):
        ystar = float(func_equilibrio(z))
        denom = ystar - z
        if abs(denom) < 1e-12:
            return 1e12       # Singularidad cercana: penalizar
        return 1.0 / denom

    error_est = None
    if esquema == 'quad':
        valor, error_est = integrate.quad(
            integrando, zf, z0, epsabs=1e-9, epsrel=1e-9, limit=200
        )
    elif esquema == 'trapz':
        pts = np.linspace(zf, z0, 2000)
        valor = integrate.trapezoid([integrando(p) for p in pts], pts)
    elif esquema == 'simpson':
        pts = np.linspace(zf, z0, 2001)   # npts impar
        valor = integrate.simpson([integrando(p) for p in pts], x=pts)
    else:
        raise ValueError(f"Esquema '{esquema}' no reconocido. Use 'quad', 'trapz' o 'simpson'.")

    razon   = np.exp(valor)
    frac_v  = 1.0 - 1.0 / razon
    frac_l  = 1.0 / razon

    return {
        'integral'           : valor,
        'razon_L0_Lf'        : razon,
        'fraccion_vaporizada': frac_v,
        'fraccion_liquida'   : frac_l,
        'error_estimado'     : error_est
    }

print(es("Función 'calcular_integral_separacion' definida"))
```

### Problema Inverso: Composición Final dado un Porcentaje de Vaporización

En muchos problemas de diseño, la variable **especificada** es la fracción de carga que se desea vaporizar, y la incógnita es la composición del líquido al final del proceso.

Se utiliza el método de Brent, que combina bisección con interpolación y es más robusto que Newton-Raphson para funciones no lineales.

```python
def hallar_composicion_final(z0, fraccion_objetivo, func_equilibrio, tol=1e-8):
    """
    Encuentra la composición final zf tal que la fracción vaporizada sea la deseada.

    Parámetros
    ----------
    z0               : float   →  composición inicial
    fraccion_objetivo: float   →  fracción de L0 que se desea vaporizar (0 < f < 1)
    func_equilibrio  : función →  y*(z)
    tol              : float   →  tolerancia de convergencia

    Retorna
    -------
    zf : float  →  composición final del líquido
    """
    if not 0.0 < fraccion_objetivo < 1.0:
        raise ValueError("La fracción debe estar estrictamente entre 0 y 1")

    ln_objetivo = np.log(1.0 / (1.0 - fraccion_objetivo))   # = ln(L0/Lf)

    def residuo(zf_prueba):
        if zf_prueba <= 0.0 or zf_prueba >= z0:
            return np.sign(zf_prueba - z0/2) * 1e6
        res = calcular_integral_separacion(z0, zf_prueba, func_equilibrio)
        return res['integral'] - ln_objetivo

    # Buscar el cero en el intervalo (1e-4, z0 - 1e-6) con método de Brent
    zf_sol = brentq(residuo, 1e-4, z0 - 1e-6, xtol=tol, maxiter=500)
    return zf_sol

print(es("Función 'hallar_composicion_final' definida"))
```

### Perfil Dinámico del Proceso

Esta función genera la evolución de todas las variables de interés a lo largo del proceso, desde la composición inicial hasta la final.

```python
def perfil_destilacion_batch(z0, zf_final, func_equilibrio, n_pasos=60):
    """
    Calcula el perfil completo de la destilación: composiciones, fracciones y balance.

    Parámetros
    ----------
    z0            : float   →  composición inicial del líquido
    zf_final      : float   →  composición final deseada
    func_equilibrio : función
    n_pasos       : int     →  resolución del perfil (número de puntos)

    Retorna
    -------
    df : DataFrame con columnas:
        z_L          : composición del líquido en el calderín
        y_star       : composición del vapor instantáneo
        y_destilado  : composición acumulada del destilado
        L_sobre_L0   : fracción de líquido remanente
        frac_vap     : fracción acumulada vaporizada
    """
    z_array = np.linspace(z0, zf_final, n_pasos)
    registros = []

    for i, z in enumerate(z_array):
        ystar = float(func_equilibrio(z))
        if i == 0:
            L_L0 = 1.0
            fv   = 0.0
            yd   = ystar           # en t=0 el destilado es el primer vapor
        else:
            res  = calcular_integral_separacion(z0, z, func_equilibrio)
            L_L0 = res['fraccion_liquida']
            fv   = res['fraccion_vaporizada']
            # Balance global: L0*z0 = Lf*zf + D*yD  →  yD = (z0 - L_L0*z) / fv
            yd   = (z0 - L_L0 * z) / fv if fv > 1e-10 else ystar

        registros.append({
            'z_L'        : float(z),
            'y_star'     : float(ystar),
            'y_destilado': float(yd),
            'L_sobre_L0' : float(L_L0),
            'frac_vap'   : float(fv)
        })

    return pd.DataFrame(registros)

print(es("Función 'perfil_destilacion_batch' definida"))
```

### Función de Visualización

```python
def graficar_analisis_batch(df_perfil, x_eq_data=None, y_eq_data=None, nombre_sistema=''):
    """
    Genera cuatro gráficos que resumen el análisis de la destilación batch.

    Subgráficos:
      1. Diagrama de equilibrio x-y con la trayectoria del proceso.
      2. Evolución de composiciones vs fracción vaporizada.
      3. Líquido remanente (L/L0) vs fracción vaporizada.
      4. Función integrando 1/(y* - z) vs z.
    """

    # Validación básica de columnas necesarias
    columnas_necesarias = ['z_L', 'y_star', 'frac_vap', 'y_destilado', 'L_sobre_L0']

    for col in columnas_necesarias:
        if col not in df_perfil.columns:
            raise ValueError(f"Falta la columna requerida en df_perfil: '{col}'")

    fig, axes = plt.subplots(2, 2, figsize=(16, 12))

    titulo = f'Análisis Destilación Batch{" — " + nombre_sistema if nombre_sistema else ""}'
    fig.suptitle(titulo, fontsize=17, y=0.99)

    # ─────────────────────────────────────────────
    # Gráfico 1: Diagrama x-y
    # ─────────────────────────────────────────────
    ax = axes[0, 0]

    if x_eq_data is not None and y_eq_data is not None:
        ax.plot(
            x_eq_data,
            y_eq_data,
            'b-',
            lw=2.5,
            label='Curva de equilibrio'
        )

    ax.plot(
        df_perfil['z_L'],
        df_perfil['y_star'],
        'ro-',
        ms=5,
        lw=1.5,
        label='Trayectoria del proceso'
    )

    ax.plot(
        [0, 1],
        [0, 1],
        'k--',
        alpha=0.45,
        label='Diagonal y = z'
    )

    ax.plot(
        df_perfil['z_L'].iloc[0],
        df_perfil['y_star'].iloc[0],
        'g^',
        ms=12,
        zorder=5,
        label='Inicio'
    )

    ax.plot(
        df_perfil['z_L'].iloc[-1],
        df_perfil['y_star'].iloc[-1],
        'rs',
        ms=12,
        zorder=5,
        label='Final'
    )

    ax.set_xlabel(r'$z_L$ (líquido)')
    ax.set_ylabel(r'$y^*$ (vapor)')
    ax.set_title('Diagrama de Equilibrio y Trayectoria')
    ax.legend(fontsize=10)
    ax.set_xlim(-0.03, 1.03)
    ax.set_ylim(-0.03, 1.03)
    ax.grid(True, alpha=0.3)

    # ─────────────────────────────────────────────
    # Gráfico 2: Composiciones vs fracción vaporizada
    # ─────────────────────────────────────────────
    ax = axes[0, 1]

    ax.plot(
        df_perfil['frac_vap'],
        df_perfil['z_L'],
        'b-',
        lw=2.5,
        label=r'$z_L$ (líquido en calderín)'
    )

    ax.plot(
        df_perfil['frac_vap'],
        df_perfil['y_star'],
        'r--',
        lw=2.5,
        label=r'$y^*$ (vapor instantáneo)'
    )

    ax.plot(
        df_perfil['frac_vap'],
        df_perfil['y_destilado'],
        'g-.',
        lw=2.5,
        label=r'$\bar{y}_D$ (destilado acumulado)'
    )

    ax.set_xlabel('Fracción vaporizada')
    ax.set_ylabel('Composición (fracción molar)')
    ax.set_title('Evolución de Composiciones')
    ax.legend()
    ax.grid(True, alpha=0.3)

    # ─────────────────────────────────────────────
    # Gráfico 3: Líquido remanente
    # ─────────────────────────────────────────────
    ax = axes[1, 0]

    ax.plot(
        df_perfil['frac_vap'],
        df_perfil['L_sobre_L0'],
        'g-',
        lw=3
    )

    ax.fill_between(
        df_perfil['frac_vap'],
        0,
        df_perfil['L_sobre_L0'],
        alpha=0.25,
        color='green'
    )

    ax.set_xlabel('Fracción vaporizada')
    ax.set_ylabel(r'$L/L_0$')
    ax.set_title('Líquido Remanente en el Calderín')
    ax.set_ylim(0, 1.05)
    ax.grid(True, alpha=0.3)

    # ─────────────────────────────────────────────
    # Gráfico 4: Integrando de Rayleigh
    # ─────────────────────────────────────────────
    ax = axes[1, 1]

    z_plot = df_perfil['z_L'].values
    ystar_p = df_perfil['y_star'].values

    denominador = ystar_p - z_plot

    # Evitar división por cero o valores negativos no físicos
    denominador_seguro = np.where(denominador > 1e-8, denominador, np.nan)

    integrando = 1.0 / denominador_seguro

    # Limitar valores demasiado grandes para que la gráfica sea legible
    integrando = np.clip(integrando, 0, 25)

    ax.plot(
        z_plot,
        integrando,
        'm-',
        lw=2.5
    )

    ax.fill_between(
        z_plot,
        0,
        integrando,
        alpha=0.25,
        color='purple'
    )

    ax.set_xlabel(r'$z_L$')
    ax.set_ylabel(r'$1/(y^* - z_L)$')
    ax.set_title('Función Integrando de la Integral de Separación')
    ax.set_ylim(0, 25)
    ax.grid(True, alpha=0.3)

    plt.tight_layout()
    plt.show()


print(es("Función 'graficar_analisis_batch' definida correctamente"))
```
